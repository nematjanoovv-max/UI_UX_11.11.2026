### Heart Disease Risk Prediction

Introduction

In this project, we aim to predict the likelihood of heart disease in individuals using various clinical, lifestyle, and demographic features such as age, gender, smoking status, cholesterol level, blood pressure, physical activity, and BMI.

Heart disease is a leading cause of death worldwide, often resulting from long-term health and lifestyle factors. Early prediction of cardiovascular risk allows for timely intervention, personalized treatment plans, and better overall health outcomes. By leveraging data science and machine learning, we can identify individuals at high risk and assist healthcare professionals in making informed decisions.

We will perform the following steps in this notebook:
- Load and preprocess the dataset
- Explore and visualize key patterns and trends
- Train and evaluate classification modelss


Let’s get started 

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# import the necessary libraries
```

```python
Data = pd.read_csv('/kaggle/input/heart-disease-risk-prediction-dataset/heart_disease_risk_dataset_earlymed.csv')
Data.head()

# load the dataset
```

```python
Data.isnull().sum().sum()

# check for missing values  
```

```python
Data.duplicated().sum()

# check for duplicates values
```

```python
Data.drop_duplicates(inplace=True)
```

```python
Data.info()
```

```python
plt.figure(figsize=(12,8))
sns.histplot(x='Age',data=Data)
plt.title('Distribution of Age')
plt.show()

# explore the data with histplot
```

```python
Data.hist(figsize=(15,15))
plt.show()

# explore the data with histgram 
```

```python
plt.figure(figsize=(15,15))
sns.heatmap(Data.corr(),annot=True)
plt.show()

# explore the data with heatmap 
```

```python
x = Data.drop(columns=['Heart_Risk'])
y = Data['Heart_Risk']
```

```python
from sklearn.model_selection import train_test_split

x_train , x_test , y_train , y_test = train_test_split(x,y,test_size=0.3,random_state=42)

# split the data
```

```python
from sklearn.linear_model import LogisticRegression

Model = LogisticRegression()
Model.fit(x_train,y_train)

# train the model
```

```python
y_pred = Model.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test,y_pred)
print(f"""The Accuracy : {accuracy * 100:.2f}%""")

# evaluate the model with accuracy score
```

```python
print("Train Accuracy:", Model.score(x_train, y_train))
print("Test Accuracy:", Model.score(x_test, y_test))
```
